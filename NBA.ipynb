```python
# Creating a New Column
df['Weight(Kgs)'] = df['Weight(lbs)'] * 0.45359237
```

```python
# @title Q10
# What is the weight in KGs for 112th record (111th Index)?
df.loc[111, 'Weight(Kgs)']
```

```python
df['Height(Mts)'] = df['Height(ft)'] / 3.28084
```

```python
# Calculate BMI of the players
df['BMI'] = df['Weight(Kgs)'] / (df['Height(Mts)'] ** 2)
```

```python
'''
Create a new columns with the name "BMI Evaluation".
Assign the BMI status based on the following condition,
BMI <= 18.4 - "Underweight"
BMI <= 24.9 - "Healthy"
BMI <= 29.9 - "Over Weight"
BMI <= 34.9 - "Severly Over Weight"
BMI <= 39.9 - "Obese"
BMI > 39.9 - "Severly Obese"
'''
# Create a empty list to store categories
BMI_Category = []

# Loop through BMI values
for bmi in df['BMI']:
  if bmi <= 18.4:
    BMI_Category.append("Underweight")
  elif bmi <= 24.9:
    BMI_Category.append("Healthy")
  elif bmi <= 29.9:
    BMI_Category.append("Overweight")
  elif bmi <= 34.9:
    BMI_Category.append("Severly Overweight")
  elif bmi <= 39.9:
    BMI_Category.append("Obese")
  else:
    BMI_Category.append("Severly Obese")


df['BMI_Category'] = BMI_Category
```

```python
# @title Q11
# Check the count of different categories in BMI Category
df['BMI_Category'].value_counts().reset_index()
```

```python
# @title Q12
# Find the total salary spent by each team on thier players
df.groupby(by = 'Team')['Salary'].sum().reset_index()
```

```python
# @title Q13
# Find the Different BMI Category count based on the Team
df.groupby(by = ['Team','BMI_Category'])['BMI'].count()
```
